```python
from langchain.chat_models import init_chat_model 
from langchain_core.prompts import PromptTemplate 
from langchain_core.output_parsers import StrOutputParser
from dotenv import load_dotenv 

load_dotenv()

model = init_chat_model(model_provider="google_genai", 
                        model ="gemini-2.5-flash",
                        temperature=0.7,
                        max_output_tokens=1024,
                        top_p=0.95,
                        top_k=40,
                        timeout = 30,
                        max_retries = 3,
                        load_env=True)
```

```python
prompt1 = PromptTemplate(
    template = "Generate a 100 words detailed report on {topic}",
    input_variables = ['topic']
)

prompt2 = PromptTemplate(
    template = "Generate 5 important point from this report  {text}",
    input_variables = ['text']
)

parser = StrOutputParser()
```

```python
chain = prompt1 | model | prompt2 | model | parser 
```

```python
result = chain.invoke({"topic": "Artificial Intelligence"})
```

```python
chain.get_graph().print_ascii()
```
