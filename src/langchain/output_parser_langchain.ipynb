```python
from langchain.chat_models import init_chat_model 
from langchain_core.prompts import PromptTemplate
from langchain_core.output_parsers import StrOutputParser , JsonOutputParser
from dotenv import load_dotenv

load_dotenv()

model = init_chat_model(model_provider="google_genai", 
                        model ="gemini-2.5-flash",
                        temperature=0.7,
                        max_output_tokens=1024,
                        top_p=0.95,
                        top_k=40,
                        timeout = 30,
                        max_retries = 3,
                        load_env=True)
```

```python
prompt1 = PromptTemplate(
    template = "Write a detail note on {topic}",
    input_variables = ['topic']
)

prompt2 = PromptTemplate(
    template = 'Write a five line report on the following text. /n {text}',
    input_variables = ['text']
)
```

### Using string output parser 

```python
parser = StrOutputParser()
```

### Using Json output parser 

```python
parser_2 = JsonOutputParser()
```

```python
prompt3 = PromptTemplate(
    template = "Give me name, description and city of a marvel superhero \n {format_instructions}",
    input_variables = [],
    partial_variables = {'format_instructions': parser_2.get_format_instructions() }
)
```

```python
chain = prompt1 | model | prompt2 | model | parser  
```

```python
result = chain.invoke({"topic": "Artificial Intelligence"})
```

```python
chain_2 = prompt3 | model | parser_2
result = chain_2.invoke({})
```
