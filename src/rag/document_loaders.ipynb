```python
from langchain.chat_models import init_chat_model 
from langchain_community.document_loaders import TextLoader , PyPDFLoader
from langchain_core.prompts import PromptTemplate
from langchain_core.output_parsers import StrOutputParser
from dotenv import load_dotenv

load_dotenv()

model = init_chat_model(
    model_provider="groq",
    model="openai/gpt-oss-20b",
    temperature=0.7,
    max_tokens=1024,
    timeout=30,
    max_retries=3,
)
```

```python
loader_text = TextLoader("/home/lakshay/LangChain/src/rag/data/turning.txt", encoding="utf8")
text_doc = loader_text.load()
```

```python
loader_pdf = PyPDFLoader("/home/lakshay/LangChain/src/rag/data/Alan Mathison Turing.pdf")
pdf_doc = loader_pdf.load()
pdf_doc[0].page_content
```

```python
prompt = PromptTemplate(
    template = "Give 5 important points from the following text. /n {text}",
    input_variables = ['text']
)
```

```python
parser = StrOutputParser()
```

```python
chain = prompt | model | parser  
```

```python
result = chain.invoke(text_doc[0].page_content)
```
